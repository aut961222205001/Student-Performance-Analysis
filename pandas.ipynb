```python
import pandas as pd

df = pd.read_csv("students.csv")
df = df.drop_duplicates()
score_columns = ["Math_Score", "Science_Score"]
for columnm in score_columns:
    df[columnm] = df[columnm].fillna(df[columnm].median())
    low = df[columnm].quantile(0.25)
    high = df[columnm].quantile(0.75)
    df[columnm] = df[columnm].clip(lower=low, upper=high)
df["Average_Score"] = df[score_columns].mean(axis=1)
print(df[["Name", "Math_Score", "Science_Score", "Average_Score"]])  # Python Pandas – Data Cleaning, Outlier Handling and Average Score

# Code	Description
#pd.read_csv()	Reads CSV file
#drop_duplicates()	Removes duplicate rows
#score_columns	Selects Math and Science columns
#for column in score_columns	Loops through both columns
#fillna()	Fills missing values
#median()	Finds middle value
#quantile(0.25)	Calculates first quartile (Q1)
#quantile(0.75)	Calculates third quartile (Q3)
#clip()	Limits values to the specified lower and upper bounds
#mean(axis=1)	Calculates row-wise average
#print()	Displays selected columns

#Important concepts
#1. Data Cleaning: Removes duplicate records and fills missing scores.
#2. Outlier Handling: Uses Q1 and Q3 to limit extreme score values.
#3. Feature Creation: Creates a new column called Average_Score.
#4. Data Display: Shows student name, Math score, Science score and average score.
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, MinMaxScaler
from sklearn.ensemble import RandomForestRegressor

# Load dataset
model_data = pd.read_csv("students.csv").dropna(subset=["Math_Score"])

# Separate input and output
X = model_data[["Science_Score", "Attendance_Percent", "City", "Club"]]
y = model_data["Math_Score"]

# Split training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Numerical preprocessing
number_steps = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler()
)

# Categorical preprocessing
category_steps = make_pipeline(
    SimpleImputer(strategy="most_frequent"),
    OneHotEncoder(handle_unknown="ignore")
)

# Combine preprocessing steps
prepare = ColumnTransformer([
    ("numbers", number_steps, ["Science_Score", "Attendance_Percent"]),
    ("categories", category_steps, ["City", "Club"])
])

# Create model
model = make_pipeline(
    prepare,
    RandomForestRegressor(random_state=42)
)

# Train model
model.fit(X_train, y_train)

# Predict Math Scores
print("Predicted Math scores:", model.predict(X_test))

# Display actual scores
print("Actual Math scores:", list(y_test))
```

```python
import pandas as pd

df = pd.read_csv("students.csv")

scores = df["Math_Score"].dropna()

q1 = scores.quantile(0.25)
q3 = scores.quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

possible_outliers = scores[
    (scores < lower_limit) | (scores > upper_limit)
]

print("Q1:",q1)
print("Q3:",q3)
print("IQR:",iqr)
print("Lower_limit",lower_limit)
print("Upper_limit:",upper_limit)
print("Possible outlier Scores:",possible_outliers.tolist())
```
