```python
import pandas as pd
df = pd.read_csv("unique_foods.csv")
result = df.groupby("Category").agg(
    Average_Price=("Price_INR", "mean"),
    Maximum_Price=("Price_INR", "max"),
    Minimum_Price=("Price_INR", "min")
)

print(result)                           # agg() - Performs multiple aggregate calculations.
```
